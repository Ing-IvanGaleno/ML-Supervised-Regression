TIPOS DE GRAFICOS DE ACUERDO A LA VARIABLE:
***Variable Cualitativa (Categorica)
-De Barras (Vertical, Horizaontal)
-De Pastel (Pie, Torta, Circular, De Sectores)

***Variable Cuantitativa (Numerica)
-Histograma de Frec.
-Linea (Series temporales)
-Área
-Boxplot  (Importantisimo para penalizar y regularizar modelos)
-Scatterplot (Dispersion, XY, Nube)    Multivariable

```python
len(df["EDAD"].unique())     #El valor obtenido representa el numero de clases (k)
#Una variable numerica es continua cuando su numero de clases (k) es grande >= 15
#Una variable numerica es discreta cuando su numero de clases (k) es grande < 15
edad = df["EDAD"].value_counts().sort_index()      #Devuelve la tabla de frecuencias absolutas (fi)
```
