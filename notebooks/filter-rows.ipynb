```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path="../data/Obesidad_Infantil.csv"
encoding= "latin1"

df=pd.read_csv(path, encoding=encoding)
```

```python
#shape Serie       (r, )    1dimension      []
#shape dataframe   (r,c)    2dimensiones    [[]]
display(df.sample(10))
```

```python
niñas= df[df["SEXO"] == "F"]
niños= df[df["SEXO"] == "M"]
#niños= df[df["SEXO"] != "F"]
```

```python
plt.figure(figsize=(10,6))
niñas["PESO (KG)"].plot.hist(label="niñas", color="pink", alpha=0.8, edgecolor= "black", bins=20)
niños["PESO (KG)"].plot.hist(label="niños", color="blue", alpha=0.8, edgecolor= "black", bins=15)
plt.ylabel("Frecuencia observada")
plt.xlabel("Pesos")
plt.legend()
```

```python
plt.figure(figsize=(10,6))
niñas["PESO (KG)"].plot.kde(label="niñas", color="red", linewidth=1.5)
niños["PESO (KG)"].plot.kde(label="niños", color="blue",linewidth=1.5)
plt.ylabel("Frecuencia observada")
plt.xlabel("Pesos")
plt.legend()
```
