```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path= "../data/Obesidad_Infantil.csv"
encoding= "latin1"
df = pd.read_csv(path, encoding=encoding)
display(df.sample(5))
```

```python
#Graficar relacion de variables numericas
sns.relplot(data=df, x="PESO (KG)", y="TALLA (MT)")
plt.show()
```

```python
#Graficar relacion de variables numericas
sns.relplot(data=df, x="CIRCUNFERENCIA DE CINTURA", y="IMC")
plt.show()
```

```python
#Graficar relacion de variables numericas
sns.scatterplot(data=df, x="CIRCUNFERENCIA DE CINTURA", y="TALLA (MT)")
plt.show()
```

```python
#Graficar relacion de variables numericas
sns.scatterplot(data=df, x="CIRCUNFERENCIA DE CINTURA", y="PESO (KG)")
plt.show()
```

```python
df_numerico = df.select_dtypes(include=np.number)
df_numerico.columns
```

```python
matriz_corr =df_numerico.corr()         #Calcula el coeficiente de correlacion indicado
display(matriz_corr)
```

```python
#Mapa de Calor o HeatMap para visualizar la intensidad de las relaciones
sns.heatmap(
    matriz_corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1.0,
    vmax=1.0,
    center=0,
    linewidth=0.5
    )
plt.title("Mapa de Calor de las correlaciones")
plt.savefig("Heatmap")
plt.show()
```
